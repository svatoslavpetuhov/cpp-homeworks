```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import HoverTool
import numpy as np

# Если работаем в Jupyter Notebook, включаем вывод в блокнот:
output_notebook()

# 1. Готовим данные
days = np.arange(1, 31)
stock_prices = 100 + np.cumsum(np.random.normal(0, 2, 30)) # Случайное блуждание курса

# 2. Создаем базовый холст
p = figure(
    title="Динамика курса акций (Интерактивный график Bokeh)", 
    x_axis_label='День', 
    y_axis_label='Цена ($)',
    width=700, 
    height=400
)

# 3. Рисуем линию и добавляем точки поверх неё
p.line(days, stock_prices, line_width=2, color="navy")
p.circle(days, stock_prices, size=8, fill_color="white", line_color="navy")

# 4. Настраиваем инструмент интерактивных подсказок (HoverTool)
# @x и @y автоматически подтянут координаты точек
hover = HoverTool(tooltips=[
    ("День эксперимента", "@x"),
    ("Цена акции", "@y{0.00}$") # Форматируем до 2 знаков после запятой
])
p.add_tools(hover)

# 5. Показываем график
show(p)
```
